Finding Missing Data

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
```

```python
 df = pd.DataFrame(data)
```

```python
df
```

```python
pd.isna(df)
```

```python
pd.isna(df).sum()
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

Removing Missing Data

```python
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3)
```

```python
df
```

Filling the missing Data

```python
df.fillna(0)
```

```python
df.fillna(df.mean())
```

```python
df
```

```python
values = {'A': 0, 'B': 400, 'C': 500, 'D': 200}
df.fillna(value= values)
```
