** Pandas Series
A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index **

CREATING SERIES

```python
import numpy as np 
import pandas as pd
```

```python
labels = ['a','b','c','d']

my_list = [1,2,3,4]

my_arr = np.array([2,8,9,0])
```

```python
pd.Series(my_list, labels)
```

```python
pd.Series(my_arr)
```

```python
d = {1:10,2:20,3:30}
```

```python
pd.Series(d)
```
